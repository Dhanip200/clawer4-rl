## 4. Load the robot on the GPU and let it stand for 1 second
The torso height should stay around **0.10 m**. If it's near 0, the robot fell through the floor.

```
import mujoco, jax
from mujoco import mjx

m = mujoco.MjModel.from_xml_path('clawer4/model/clawer4.xml')
d = mujoco.MjData(m)
mx = mjx.put_model(m)
dx = mjx.put_data(m, d)
step = jax.jit(mjx.step)
for _ in range(500):  # 500 x 0.002 s = 1 s
    dx = step(mx, dx)
print('device:', jax.devices())
print('torso height after 1 s:', float(dx.qpos[2]), 'm')
```

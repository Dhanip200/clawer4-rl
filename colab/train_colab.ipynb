## 3. Train (~55 min on a T4) and make a video

For a quick 7-minute check first, use `train_policy(out='practice.params', num_timesteps=500_000, num_envs=256, batch_size=64, num_evals=3)`.

```
import os, sys
os.environ['MUJOCO_GL'] = 'egl'
sys.path.insert(0, 'clawer4/rl')
from train import train_policy, make_video

make_policy, params = train_policy()
make_video(make_policy, params, command=(0.2, 0, 0), seconds=10, out='walk.mp4')
```
